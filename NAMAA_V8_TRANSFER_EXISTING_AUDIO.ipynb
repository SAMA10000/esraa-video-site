# Transfer verified synthetic Egyptian narration
Finds existing mp3; if absent, downloads just the MP3 from the saved Kaggle Version 1; then makes a temporary file transfer with expiry 60 minutes. Does not rerun TTS. No personal data, just synthetic audio.

```python
import os, json, mimetypes
from pathlib import Path
import requests
p=Path('/kaggle/working/namaa_mansa_60s.mp3')
print('DIRECT_MP3_EXISTS',p.exists(),flush=True)
if not p.exists():
    import kagglehub
    print('FETCHING_SAVED_NOTEBOOK_VERSION_1',flush=True)
    dst=kagglehub.notebook_output_download('fashionforyou/notebooke0df344af2/versions/1', path='namaa_mansa_60s.mp3')
    p=Path(dst)
    print('VERSION1_PATH',str(p),flush=True)
assert p.exists() and 100000<p.stat().st_size<2000000, ('MP3_NOT_FOUND_OR_WRONG_SIZE',str(p),p.stat().st_size if p.exists() else 0)
print('MP3_SIZE',p.stat().st_size,flush=True)
with p.open('rb') as f:
    r=requests.post('https://tmpfiles.org/api/v1/upload',files={'file':(p.name,f,'audio/mpeg')},timeout=55)
print('TRANSFER_HTTP_STATUS',r.status_code,flush=True)
print('TRANSFER_RESPONSE',r.text[:3000],flush=True)
r.raise_for_status()
data=r.json()
assert data.get('status')=='success',data
print('UPLOAD_SUCCESS_URL',data['data']['url'],flush=True)
```
